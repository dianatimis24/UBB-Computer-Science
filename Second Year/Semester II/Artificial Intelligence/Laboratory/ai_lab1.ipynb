_**Tensor Shapes**_

On performing operations on two or more tensors, they will need to be of the same shape - that is, having the same number of dimensions and the same number of cells in each dimension. For that, we have the torch.*_like() methods:

```python
x = torch.empty(2, 2, 3)
print(x.shape)
print(x)

empty_like_x = torch.empty_like(x)
print(empty_like_x.shape)
print(empty_like_x)

zeros_like_x = torch.zeros_like(x)
print(zeros_like_x.shape)
print(zeros_like_x)

ones_like_x = torch.ones_like(x)
print(ones_like_x.shape)
print(ones_like_x)

rand_like_x = torch.rand_like(x)
print(rand_like_x.shape)
print(rand_like_x)
```

## Task 6 -- An application: Simulate an unfair probability spinner

Consider an unfair probability spinner with n slots. Each slot has a probability P_i ∈ [0,1], where i ∈ {0,1,...,n-1}, and ∑(i=0, n−1)P_i = 1.

Write a function that randomly generates an index i with the probability P_i from the set {0,1,...,n-1}.

```python
def spinner(probability_distribution):
    return np.random.choice(len(probability_distribution), p=probability_distribution)
```

```python
probability_distribution = np.array([0.1, 0.2, 0.3, 0.4])
spinner(probability_distribution)
```

The same problem but instead of a distribution of probability the input is a list of natural numbers that holds how many degrees has the center angle θ_i corespondent to index i.

### Example:

Consider we have a wheel with 4 slots:



```python
def transform_angles(angles):
    return np.array(angles) / sum(angles)
```

```python
angles = np.array([45, 85, 70, 160])
colors = ["red", "yellow", "green", "blue"]
colors[spinner(transform_angles(angles))]
```

## Task 7 -- Determine some outcomes using Monte Carlo Simulations

Consider a large pipe with a radius R=30cm that has the opening pointing straight up, as in the figure. Around this pipe is a rectangular wall with an edge l=1m. Balls of radius r=1cm fall down uniformly from above inside the square yard. Compute, using a Monte Carlo simulation, the probability that a ball falls inside the pipe. Compare the results with the mathematically deduced probability.



```python
RADIUS = 0.3
LENGTH = 1
BALL_RADIUS = 0.01
trials = [10, 100, 1000, 10000, 100000, 1000000, 10000000]

def generate_trials(number_of_trials):
    x = np.random.uniform(-LENGTH / 2, LENGTH / 2, number_of_trials)
    y = np.random.uniform(-LENGTH / 2, LENGTH / 2, number_of_trials)

    distances = np.sqrt(x**2 + y**2)
    inside = distances < RADIUS

    count = np.sum(inside)
    return count / number_of_trials

def compute_probabilities():
    for number_of_trials in trials:
        print("number of trials = %d, estimate = %f" % (number_of_trials, generate_trials(number_of_trials)))


print("Exact probability: ", np.pi * (RADIUS - BALL_RADIUS)**2, "\n")
compute_probabilities()
```
